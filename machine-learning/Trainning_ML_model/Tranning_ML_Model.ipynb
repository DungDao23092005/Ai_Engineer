### 5.Model Evaluation

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
import pandas as pd

# Make predictions
rf_preds = rf_model.predict(X_valid)

# Evaluate Random Forest model
rf_mae = mean_absolute_error(y_valid, rf_preds)
rf_rmse = root_mean_squared_error(y_valid, rf_preds)
rf_r2 = r2_score(y_valid, rf_preds)

# Show results
rf_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2 Score"],
    "Random Forest": [rf_mae, rf_rmse, rf_r2]
})

rf_results
```

### Model Evaluation Result

The Random Forest model achieved the following results:

- MAE: 24069.39
- RMSE: 38598.84
- R2 Score: 0.7843

The R2 Score of 0.7843 means that the model can explain about 78.43% of the variation in house prices.  
The MAE value shows that the model's predictions are off by around 24,069 on average.

Overall, the Random Forest model performs reasonably well, but the prediction accuracy can be improved by using more features, feature engineering, or hyperparameter tuning.
